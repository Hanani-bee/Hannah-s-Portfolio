<h4><strong>Laboratory Task 2</strong></h4>

<b>Instruction:</b> Perform a single forward pass and compute for the error.

$$
x =
\begin{bmatrix}
1 \\
0 \\
1
\end{bmatrix}
$$

$$
y =
\begin{bmatrix}
1
\end{bmatrix}
$$

$$
f = \max(0, Z_n)
$$

$$
\text{Hidden unit weights} =
\begin{bmatrix}
w_{11} & w_{12} \\
w_{13} & w_{14} \\
w_{15} & w_{16}
\end{bmatrix}
=
\begin{bmatrix}
0.2 & -0.3 \\
0.4 & 0.1 \\
-0.5 & 0.2
\end{bmatrix}
$$

$$
\text{Output unit weights} =
\begin{bmatrix}
w_{21} \\
w_{22}
\end{bmatrix}
=
\begin{bmatrix}
-0.3 \\
-0.2
\end{bmatrix}
$$

$$
\theta =
\begin{bmatrix}
\theta_1 \\
\theta_2 \\
\theta_3
\end{bmatrix}
=
\begin{bmatrix}
-0.4 \\
0.2 \\
0.1
\end{bmatrix}
$$

<h5> Output: </h5>

````python
import numpy as np

x = np.array([1, 0, 1])          # input
y = 1                             # target
theta1, theta2, theta3 = -0.4, 0.2, 0.1

W1 = np.array([[0.2, -0.3],       # hidden weights (3x2)
               [0.4,  0.1],
               [-0.5, 0.2]])
W2 = np.array([-0.3, -0.2])       # output weights (2x1)

relu = lambda z: np.maximum(0, z)  # f = max(0, Zn)

# forward pass
h = relu(x @ W1 + [theta1, theta2])   # hidden layer output
o = relu(h @ W2 + theta3)             # network output

# error
error = 0.5 * (y - o) ** 2            # squared error

print("output:", o, "error:", error)
````

<h4><strong>Laboratory Task 3</strong></h4>

<b>Instruction:</b> Perform a forward and backward propagation in python using the inputs from <b>Laboratory Task 2</b>

```python
x = np.array([1, 0, 1])
y = np.array([1])

# use relu as the activation function.

# learning rate
lr = 0.001
```

<h5> Output: </h5>

````python
lr = 0.001                        # learning rate

# forward propagation 
z_hidden = x @ W1 + [theta1, theta2]
h = relu(z_hidden)
z_out = h @ W2 + theta3
o = relu(z_out)
error = 0.5 * (y - o) ** 2

# backward propagation
relu_deriv = lambda z: (z > 0).astype(float)

delta_o = o - y                                  # output error signal
delta_h = (delta_o * W2) * relu_deriv(z_hidden)  # hidden error signal

dW2 = h * delta_o                # gradient w.r.t. output weights
dW1 = np.outer(x, delta_h)       # gradient w.r.t. hidden weights

# update weights
W2 -= lr * dW2
theta3 -= lr * delta_o
W1 -= lr * dW1
theta1, theta2 = np.array([theta1, theta2]) - lr * delta_h

print("o:", o, "error:", error)
````

 <h4><strong>Laboratory Task 4</strong></h4>

 <b>Instruction:</b> Train a linear regression model in PyTorch using a regression dataset. Use the following parameters.

 <ul>
    <li>Criterion: MSE Loss</li>
    <li>Fully Connected Layers x 2</li>
    <li>Batch Size: 8</li>
    <li>Optimizer: SGD</li>
    <li>Epoch: 1000</li>
</ul>

<h5> Output: </h5>

````python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_regression

# regression dataset
X, y = make_regression(n_samples=200, n_features=1, noise=10, random_state=42)
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).view(-1, 1)

# standardize to prevent exploding gradients
X = (X - X.mean()) / X.std()
y = (y - y.mean()) / y.std()

loader = DataLoader(TensorDataset(X, y), batch_size=8, shuffle=True)

# model: 2 fully connected layers
class LinearRegression(nn.Module):
    def __init__(self, in_features):
        super().__init__()
        self.fc1 = nn.Linear(in_features, 16)
        self.fc2 = nn.Linear(16, 1)

    def forward(self, x):
        x = self.fc1(x)
        return self.fc2(x)

model = LinearRegression(X.shape[1])
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)   # lower lr avoids divergence

# training loop
for epoch in range(1000):
    for xb, yb in loader:
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)   # MSE loss
        loss.backward()
        optimizer.step()

print("final loss:", loss.item())
````
